# PHẦN B — DESCRIPTIVE STATISTICS
## Group 1 — Central Tendency

```python
# TODO
for col in ['age', 'fare','survived', 'pclass', 'parch', 'sibsp']: 
    s = df[col].dropna() 
    mean, median, mode = s.mean(), s.median(), s.mode()[0]
    print(f"{col.upper()}: Mean={mean:.2f}, Median={median:.2f}, Mode={mode:.2f}")
print()
print("Mode pclass:", df['pclass'].mode()[0])
print("Mode embarked:", df['embarked'].mode()[0])
```

## Group 2 — Dispersion

```python
# TODO
for col in ['age', 'fare']:
    s = df[col].dropna()
    range_ = s.max() - s.min()
    std_ = s.std()
    q1, q3 = s.quantile([0.25, 0.75])
    iqr = q3 - q1
    cv = std_ / s.mean()
    print(f"--- {col.upper()} ---")
    print(f"Range={range_:.2f}, Std={std_:.2f}, IQR={iqr:.2f}, CV={cv:.2f}")
```

## Group 3 — Location and Shape

```python
# TODO
for col in ['age', 'fare']:
    s = df[col].dropna()
    skew_, kurt_ = stats.skew(s), stats.kurtosis(s)
    print(f"--- {col.upper()} ---")
    print(f"P25={s.quantile(0.25):.2f}, P50={s.quantile(0.5):.2f}, P75={s.quantile(0.75):.2f}")
    print(f"Skewness={skew_:.2f}, Kurtosis={kurt_:.2f}\n")
```

## Câu hỏi 1: Hạng vé nào có tỷ lệ sống sót cao nhất, chênh lệch bao nhiêu so với hạng thấp nhất?

```python
# TODO
# TODO
tls = (df.groupby('class')['survived'].mean() * 100).sort_values(ascending=False)
print("Tỷ lệ sống sót theo hạng vé (%):")
print(tls.round(2))

chenhlech = tls.max() - tls.min()
print()
print("Chênh lệch tỷ lệ sống sót giữa hạng vé cao nhất và thấp nhất:")
print(chenhlech.round(2), "%")
print()
print("Vé hạng nhất có tỷ lệ sống sót cao nhất, trong khi vé hạng ba có tỷ lệ sống sót thấp nhất.")
```

## Câu hỏi 2: Giới tính hay hạng vé ảnh hưởng đến sống sót mạnh hơn?

```python
# TODO
pivot_titanic = df.pivot_table(
    index='class',      
    columns='sex',     
    values='survived',  
    aggfunc='mean',   
    fill_value=0     
).reset_index()

pivot_titanic[['female', 'male']] = pivot_titanic[['female', 'male']] * 100

print("Bảng tỷ lệ sống sót (%) theo Hạng vé và Giới tính:")
print(pivot_titanic.round(2))
print()
print("Giới tính có ảnh hưởng đến sống sót vì phụ nữ có tỷ lệ sống sót cao hơn nam giới trong cùng hạng vé.")
```

## Câu hỏi 3: Vé đắt hơn có thực sự sống sót cao hơn không?

```python
# TODO
print(df.groupby("survived")["fare"].mean())
print(df["fare"].corr(df["survived"]))
```

## Câu hỏi 4: Gia đình đông người có ảnh hưởng đến khả năng sống sót không?

```python
# TODO
ssgd = df.groupby('family_size')['survived'].mean() * 100
print("Tỷ lệ sống sót theo số lượng người trong gia đình (%):")
print(ssgd.round(2))
print()
print("Những người đi 2 người hoặc đi cùng gia đình nhỏ có tỷ lệ sống sót cao hơn so với những người đi cùng gia đình lớn từ 5 thành viên trở lên.")
```
